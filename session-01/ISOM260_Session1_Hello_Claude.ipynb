## 3. Your first call

Three things are happening: you create a client with your key, you send a list of
messages, and you get text back. Every agent you build this semester is this call
in a loop.

```python
from anthropic import Anthropic

client = Anthropic(api_key=api_key)

reply = client.messages.create(
    model='claude-sonnet-4-5',
    max_tokens=300,
    messages=[
        {'role': 'user',
         'content': 'In two sentences: name one task in retail operations you would '
                    'NOT trust an AI model with, and say why.'}
    ],
)

print(reply.content[0].text)
```

**Read what came back.** Do you agree with it? How would you check?

That question is the entire course. Producing text is the easy part.

## 4. Now break it

Pick a field you know better than the room — your job, your internship, your
family's business. Ask something you can verify, and push until it is confidently
wrong.

Notice what it does *not* do: it never says "I am not sure." There is no built-in
signal for its own uncertainty. Building that signal is what Sessions 7 and 10 are

```python
question = 'REPLACE ME with something from your field that you can check'

reply = client.messages.create(
    model='claude-sonnet-4-5',
    max_tokens=400,
    messages=[{'role': 'user', 'content': question}],
)

print(reply.content[0].text)
```

## 5. Put a number on it

One wrong answer is an anecdote. Five scored answers is evidence — and evidence is
what you will be asked for every week.

Fill in five questions from your field where you already know the answer, then
score what comes back by hand.

```python
questions = [
    'Question 1',
    'Question 2',
    'Question 3',
    'Question 4',
    'Question 5',
]

for i, q in enumerate(questions, 1):
    reply = client.messages.create(
        model='claude-sonnet-4-5',
        max_tokens=250,
        messages=[{'role': 'user', 'content': q}],
    )
    print(f'--- {i}. {q}')
    print(reply.content[0].text.strip())
    print()

# Score them yourself: how many were right? Did its tone change on the wrong ones?
```
